```python
# Take one string and count vowels and consonants present in it

s = 'Hello all welcome to Python Programming'
vo = 0
co = 0
s = s.lower()
# remove spaces
s = s.replace(' ','')
s
vowels = 'aeiou'

for item in s:
  if item in vowels:
    vo+=1  # increament for Vowel
  else:
    co+=1 #increament for Consonants
print('Total Vowel count is',vo)
print('Total Consonant count is',co)
```

```python
str1 = input('Enter your first string-->')
str2 = input('Enter your Second string-->')
str1 = str1.lower()
str2 = str2.lower()

if sorted(str1) == sorted(str2):
  print('String 1:',str1,'and String2:',str2,'are anagrams')
else:
  print('String 1:',str1,'and String2:',str2,'are not anagrams')
```

```python
# Check whether the given numbers is prime or not
# Prime number is number which divisible by 1 and itself
# Example-  17,7,3,13.........

num = int(input('Enter the number-->'))

if num<=1:
  print(num,'is a prime number')
else:
  is_prime = True

# for loop to identify factors
for i in range(2,int(120**0.5)+1):
  if num % i ==0:
    is_prime = False # number is not Prime


  break
if is_prime:
  print(num,'is a Prime Number')
else:
  print(num,'is not a Prime Number')

```

```python
# Check given string is Palindrome or not
# Ex: madam aba

s = input("Enter a string: ")
s = s.lower().replace(" ", "")

if s == s[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```
