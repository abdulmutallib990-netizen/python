# **Numpy Slicing and Indexing**

```python
x = np.arange(24).reshape(6,4)
x
```

```python
x[3] # using index we can access one row at a time
```

```python
print(x[3][3])  # row with elements index
```

```python
x[3,3]
```

```python
# numpy array is mutable in nature
x[3,3]
```

```python
# replace 15 by 50
x[3,3] = 50
x
```

```python
# acess more rows
x[1:4]  # [start_index:stop_index]
```

```python
# acess last 3 rows
# x[-3]  # -ve index
x[3:]  # +ve index
```

```python
x [::-2]  # stepping
```

```python
x
```

```python
# column operation

x[-2:,-2:]  # [row,column]
```

```python
x[-2:,-2]
```

```python
x [2:4,1:3]
```
