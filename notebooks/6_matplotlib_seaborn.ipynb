# **Matplotlip**

```python
import matplotlib.pyplot as plt
```

```python
# We can perform differnt types of analysis
# Univariate --> We are using single feature
# Bi-Variate---> 2 features
# Multi-Vriate-->More than 2 features(State vs population---> Gender)
```

```python
days = ['mon','tue','wed','thur','fri']
rain = [120,30,25,20,80]
```

```python
# bi-Variate plots- line, scatter,bar ,bar,

plt.plot(days,rain,'r--o',label='Daywise rainfall in mm')
plt.title('Weekly Rainfall data')
plt.xlabel('Days Mon- Fri')
plt.ylabel('rainfall in mm')
plt.grid()
plt.legend()
plt.savefig('sample_png',dpi = 400)
```

```python
plt.plot(days,rain,'r--o')
plt.grid()
plt.title('Weekly Rainfall data')
plt.xlabel('Days Mon- Fri')
plt.ylabel('Rainfall in mm')
```

```python
# bar plot

#plt.bar(days,rain,color= ['r','y','m','c','k'],label='Daywise rainfall in mm')
plt.barh(days,rain,color= ['r','y','m','c','k'],label='Daywise rainfall in mm')
plt.title('Weekly Rainfall data')
plt.ylabel('Days Mon- Fri')
plt.xlabel('Rainfall in mm')
plt.grid()
plt.legend()
```
