```python
import pandas as pd
```

```python
path = "/content/Salary_Data.csv"

path
```

```python
# load dataset

sal = pd.read_csv(path)
sal[:2]
```

```python
# prepare data

X = sal [['YearsExperience']]  # ML algorithams

Y = sal [['Salary']]
```

```python
import matplotlib.pyplot as plt

plt.scatter(X,Y,color='red',marker= '*')
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
# Training---> using fit () method
model.fit(X,Y)
```

```python
# Training ---means it is calculating the slops and in
```

```python
model.predict([[0]])
```

```python
m = model.coef_
m
```

```python
c = model.intercept_
c
```

```python
m*7.5+c
```
