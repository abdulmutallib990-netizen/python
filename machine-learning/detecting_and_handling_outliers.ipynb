```python
import numpy as np
outliers = []  #
def detect_outliers_zscore(data):
    thres = 3
    mean = np.mean(data)
    std = np.std(data)
    # print(mean, std)
    for i in data:
        z_score = (i-mean)/std
        print('Z_score',z_score,'for value:,',i)
        if (np.abs(z_score) > thres):
            outliers.append(i)
    return outliers# Driver code
sample_outliers = detect_outliers_zscore(sample)
print("Outliers from Z-scores method: ", sample_outliers)
```

```python
# Using IQR method

outliers_2 = []
def detect_outliers_iqr(data):
    data = sorted(data)
    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)
    # print(q1, q3)
    IQR = q3-q1
    lwr_bound = q1-(1.5*IQR)
    upr_bound = q3+(1.5*IQR)
    # print(lwr_bound, upr_bound)
    for i in data:
        if (i<lwr_bound or i>upr_bound): # value less than lower bound and greater than upper bound is an outlier
            outliers_2.append(i)
    return outliers_2# Driver code
sample_outliers = detect_outliers_iqr(sample1)
print("Outliers from IQR method: ", sample_outliers)
```
