# **Sentence Transformer**





```python
from sentence_transformers import SentenceTransformer
import numpy as np

model = SentenceTransformer("all-MiniLM-L6-v2")
print("Model loaded!")
```

```python
sentences = [
    "Machine learning is a subset of AI",
    "ML kmeans is algorithms to know the value",
    "marvel's domsday is releasing in december"
]

embeddings = []
for s in sentences:
    emb = model.encode("search_document: " + s)
    embeddings.append(emb)

print("Vector size:", len(embeddings[0]))
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
print("Comparing sentences:\n")
print(f"1: '{sentences[0]}'")
print(f"2: '{sentences[1]}'")
print(f"3: '{sentences[2]}'")

print(f"\n1 vs 2 (similar meaning): {cosine_similarity(embeddings[0], embeddings[1]):.4f}")
print(f"1 vs 3 (different topic): {cosine_similarity(embeddings[0], embeddings[2]):.4f}")
```

# **Open AI Embeding Model**

```python
from openai import OpenAI
from google.colab import userdata
import numpy as np

client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
print("Client ready!")
```

```python
sentences = [
    "Virat Kohli compeleted his 15000 runs milestones in ODI",
    "Virat Kohli has batted in a total of 630 international innings across all formats.",
    "Cristiano Ronaldo has scored a total of 979 official career goals for club and country."
]

embeddings = []
for s in sentences:
    res = client.embeddings.create(
        model="text-embedding-3-small",
        input=s
    )
    embeddings.append(res.data[0].embedding)
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        score = cosine_similarity(embeddings[i], embeddings[j])
        print(f"{i+1} vs {j+1}: {score:.4f}")
```

# **Nomic Embedding Model**

```python
nomic_emb = []
for s in sentences:
    emb = model.encode("search_document: " + s)
    nomic_emb.append(emb)
```

```python
import numpy as np

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("Pair      Nomic    OpenAI")
for i in range(len(sentences)):
    for j in range(i + 1, len(sentences)):
        n = cosine_similarity(nomic_emb[i], nomic_emb[j])
        o = cosine_similarity(embeddings[i], embeddings[j])
        print(f"{i+1} vs {j+1}    {n:.4f}   {o:.4f}")
```
