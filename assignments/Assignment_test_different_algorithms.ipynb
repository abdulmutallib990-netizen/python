```python
import pandas as pd

path = 'https://raw.githubusercontent.com/hakimyameen/Datasets/refs/heads/main/heart.csv'

h = pd.read_csv(path)
h[:2]
```

```python
# data prepare
X = h[['age','trestbps','chol']]
#X = h.iloc[:,:-1]
y = h['target']
```

```python
# Split the dataset

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2)
```

```python
# test on KNeighborsClassifier algorithms
from sklearn.neighbors import KNeighborsClassifier
```

```python
kc = KNeighborsClassifier()
kc.fit(X_train,y_train)
```

```python
kc_pred = kc.predict(X_test)
```

```python
kc_pred[:2]
```

```python
# everything in one form
from sklearn.metrics import classification_report

print(classification_report(y_test,kc_pred))
```

```python
# Test on GaussianNB algorithms
from sklearn.naive_bayes import GaussianNB
```

```python
gn =  GaussianNB()
gn.fit(X_train,y_train)
```

```python
gn_pred = gn.predict(X_test)
```

```python
gn_pred[:2]
```

```python
print(classification_report(y_test,gn_pred))
```

```python
# Test on RandomForestClassifier algorithm
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier()
rfc.fit(X_train,y_train)
```

```python
rfc_pred = rfc.predict(X_test)
```

```python
rfc_pred[:2]
```

```python
print(classification_report(y_test,rfc_pred))
```

```python
# Test on  SVC algorithms
from sklearn.svm import SVC
```

```python
svc = SVC()
svc.fit(X_train,y_train)
```

```python
svc_pred = svc.predict(X_test)
```

```python
print(classification_report(y_test,svc_pred))
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Assuming your trained models are: kc, svc, gn, rfc
models = {
    'KNN': kc,
    'SVM': svc,
    'Naive Bayes': gn,
    'Random Forest': rfc,
}

rows = []
for name, model in models.items():
    y_pred = model.predict(X_test)

    # Metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    # Confusion Matrix
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred).ravel()

    rows.append({
        'Model': name,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'TP': tp,
        'FP': fp,
        'FN': fn,
        'TN': tn,
    })

# Create and display
comparison_df = pd.DataFrame(rows).round(4).sort_values(by='F1-Score', ascending=False)
print(comparison_df)
```
